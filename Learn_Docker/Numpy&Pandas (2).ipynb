#### Attributes of numpy array
- type(arr) - Will tell you the datatype of the variable as a whole
- arr.shape - Will tell you the number of rows and columns (it hels us to understand the structure of the array)
- arr.ndim - Will tell you how many dimensions of array the data is
- arr.dtype - Will tell you the datatype of elements inside it
- arr.size - Will tell you the total number of elements inside the array

```python
print(arr1, type(arr1), arr1.shape, arr1.ndim, arr1.dtype, arr1.size)
```

```python
print(arr2, type(arr2), arr2.shape, arr2.ndim, arr2.dtype, arr2.size)
```

```python
print(arr3, type(arr3), arr3.shape, arr3.ndim, arr3.dtype, arr3.size)
```

### Locaters and Ilocaters
#### Locaters are name based access of rows and columns. If rows are not named, row numbers also work
- df.loc[col_name,row_name]
- df.loc[[list of col names],[list of row names]]
- df.loc[[list of col_ names] start_row:end_row]
#### Ilocaters are index based access of rows and columns. 
- df.iloc[col_index,row_name]
- df.iloc[[list of col names],[list of row names]]
- df.iloc[start_col:end_col, start_row:end_row]
